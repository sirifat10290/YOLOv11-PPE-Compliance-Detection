```python
!pip install albumentations -q

import os, cv2, random
import albumentations as A

random.seed(42)
SRC = '/content/Construction-Site-Safety-27'
BASE_OUT = '/content/aug_datasets'
os.makedirs(BASE_OUT, exist_ok=True)

def load_yolo_labels(label_path):
    boxes, classes = [], []
    if not os.path.exists(label_path):
        return boxes, classes
    with open(label_path) as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 5:
                continue
            c, x, y, w, h = parts
            classes.append(int(c))
            boxes.append([float(x), float(y), float(w), float(h)])
    return boxes, classes

def save_yolo_labels(label_path, boxes, classes):
    with open(label_path, 'w') as f:
        for c, b in zip(classes, boxes):
            f.write(f"{c} {b[0]:.6f} {b[1]:.6f} {b[2]:.6f} {b[3]:.6f}\n")

transforms = {
    'brightness': A.Compose(
        [A.RandomBrightnessContrast(brightness_limit=0.4, contrast_limit=0.2, p=1.0)],
        bbox_params=A.BboxParams(format='yolo', label_fields=['class_labels'])),
    'blur': A.Compose(
        [A.OneOf([A.MotionBlur(blur_limit=9, p=1.0), A.GaussianBlur(blur_limit=(5, 9), p=1.0)], p=1.0)],
        bbox_params=A.BboxParams(format='yolo', label_fields=['class_labels'])),
    'rotation': A.Compose(
        [A.Rotate(limit=25, border_mode=cv2.BORDER_CONSTANT, p=1.0)],
        bbox_params=A.BboxParams(format='yolo', label_fields=['class_labels'], min_visibility=0.3)),
    'noise': A.Compose(
        [A.GaussNoise(var_limit=(20.0, 60.0), p=1.0)],
        bbox_params=A.BboxParams(format='yolo', label_fields=['class_labels'])),
}

def build_variant(name, transform):
    out_img_dir = f'{BASE_OUT}/{name}/train/images'
    out_lbl_dir = f'{BASE_OUT}/{name}/train/labels'
    os.makedirs(out_img_dir, exist_ok=True)
    os.makedirs(out_lbl_dir, exist_ok=True)
    img_dir = f'{SRC}/train/images'
    lbl_dir = f'{SRC}/train/labels'
    for fname in os.listdir(img_dir):
        img_path = f'{img_dir}/{fname}'
        lbl_path = f'{lbl_dir}/{os.path.splitext(fname)[0]}.txt'
        img = cv2.imread(img_path)
        if img is None:
            continue
        boxes, classes = load_yolo_labels(lbl_path)
        try:
            out = transform(image=img, bboxes=boxes, class_labels=classes)
            aug_img, aug_boxes, aug_classes = out['image'], out['bboxes'], out['class_labels']
        except Exception:
            aug_img, aug_boxes, aug_classes = img, boxes, classes
        cv2.imwrite(f'{out_img_dir}/{fname}', aug_img)
        save_yolo_labels(f'{out_lbl_dir}/{os.path.splitext(fname)[0]}.txt', aug_boxes, aug_classes)
    for split in ['valid', 'test']:
        dst = f'{BASE_OUT}/{name}/{split}'
        if not os.path.exists(dst):
            os.symlink(f'{SRC}/{split}', dst)
    import shutil
    shutil.copy(f'{SRC}/data.yaml', f'{BASE_OUT}/{name}/data.yaml')
    print(f"Built variant: {name}")

for name, t in transforms.items():
    build_variant(name, t)

print("All 4 augmented variants ready.")
```

```python
from ultralytics import YOLO
import cv2

def compliance_check(image_path, conf=0.4):
    results = model.predict(image_path, conf=conf, verbose=False)[0]
    names = results.names
    boxes = results.boxes.xyxy.cpu().numpy()
    classes = [names[int(c)] for c in results.boxes.cls.cpu().numpy()]

    persons = [b for b, c in zip(boxes, classes) if c == 'Person']
    hardhats = [b for b, c in zip(boxes, classes) if c == 'Hardhat']
    vests = [b for b, c in zip(boxes, classes) if c == 'Safety Vest']
    masks = [b for b, c in zip(boxes, classes) if c == 'Mask']

    def overlaps(person_box, item_box, thresh=0.05):
        px1, py1, px2, py2 = person_box
        ix1, iy1, ix2, iy2 = item_box
        ox1, oy1 = max(px1, ix1), max(py1, iy1)
        ox2, oy2 = min(px2, ix2), min(py2, iy2)
        if ox2 <= ox1 or oy2 <= oy1:
            return False
        inter = (ox2 - ox1) * (oy2 - oy1)
        item_area = (ix2 - ix1) * (iy2 - iy1)
        return (inter / item_area) > thresh if item_area > 0 else False

    report = []
    for i, p in enumerate(persons):
        has_hardhat = any(overlaps(p, h) for h in hardhats)
        has_vest = any(overlaps(p, v) for v in vests)
        has_mask = any(overlaps(p, m) for m in masks)

        items_present = sum([has_hardhat, has_vest, has_mask])

        if items_present == 3:
            status = "Fully Compliant"
        elif items_present >= 1:
            status = "Partially Compliant"
        else:
            status = "Non-Compliant"

        report.append({
            "worker": i + 1,
            "hardhat": has_hardhat,
            "vest": has_vest,
            "mask": has_mask,
            "status": status
        })


    im = cv2.imread(image_path)
    color_map = {"Fully Compliant": (0, 200, 0), "Partially Compliant": (0, 165, 255), "Non-Compliant": (0, 0, 255)}
    for i, p in enumerate(persons):
        x1, y1, x2, y2 = map(int, p)
        status = report[i]["status"]
        cv2.rectangle(im, (x1, y1), (x2, y2), color_map[status], 3)
        cv2.putText(im, status, (x1, max(y1 - 10, 10)), cv2.FONT_HERSHEY_SIMPLEX, 0.8, color_map[status], 2)

    out_path = image_path.replace('.jpg', '_compliance3.jpg').replace('.png', '_compliance3.png')
    cv2.imwrite(out_path, im)
    return report, out_path
```

```python
import glob, random, os, shutil

random.seed(123)
all_images = glob.glob('/content/Construction-Site-Safety-27/valid/images/*.jpg')
sample_30 = random.sample(all_images, 30)

os.makedirs('/content/compliance_eval_batch', exist_ok=True)
summary = []

for i, img_path in enumerate(sample_30):
    try:
        report, out_path = compliance_check(img_path)
        if len(report) == 0:
            continue

        new_name = f"/content/compliance_eval_batch/img_{i+1:02d}_workers{len(report)}.jpg"
        shutil.copy(out_path, new_name)
        summary.append((new_name, len(report), [r['status'] for r in report]))
    except Exception as e:
        print(f"Skipped {img_path}: {e}")

for s in summary:
    print(s[0].split('/')[-1], "| workers:", s[1], "| statuses:", s[2])

print(f"\nTotal images processed: {len(summary)}")


shutil.make_archive('/content/compliance_eval_batch', 'zip', '/content/compliance_eval_batch')
print("Zip ready: /content/compliance_eval_batch.zip")
```
